## 📦 Extracting Top-100 Classes from ResNet-Formatted Dataset

This notebook filters out and creates a new dataset containing only the **Top-100 selected bird classes** from a full ResNet-style dataset. It works by:

### ✅ Key Steps:
1. **Define the list of 100 selected classes** (preserving original names with spaces and capitalization).
2. **Loop through the `train/` and `test/` splits** from the original dataset.
3. **Match and extract class folders** using their folder name suffixes (e.g., `019.Black_footed_Albatross` ➝ `Black footed Albatross`).
4. **Copy matched folders** into a new dataset directory (`resnet_dataset_top100`), keeping class names readable and user-friendly.

> This preprocessing step is essential to reduce the dataset size and focus on a curated subset of bird species for training or evaluation.

### 📁 Final Output Directory Structure

    resnet_dataset_top100/
    ├── train/
    │   ├── Prothonotary Warbler/
    │   ├── Yellow breasted Chat/
    │   └── ...
    └── test/
        ├── Prothonotary Warbler/
        ├── Yellow breasted Chat/
        └── ...


```python
import os
import shutil

# List of your top‑100 classes (exactly as printed, spaces and capitalization preserved)
selected_classes = [
    "Prothonotary Warbler", "Yellow breasted Chat", "Yellow headed Blackbird",
    "Purple Finch", "Horned Lark", "Red bellied Woodpecker", "Common Yellowthroat",
    "White crowned Sparrow", "Red winged Blackbird", "European Goldfinch",
    "Black throated Blue Warbler", "Cardinal", "Least Auklet", "Scarlet Tanager",
    "Rose breasted Grosbeak", "Tropical Kingbird", "Horned Grebe",
    "Vermilion Flycatcher", "Lazuli Bunting", "Myrtle Warbler",
    "Black and white Warbler", "Evening Grosbeak", "Black billed Cuckoo",
    "Parakeet Auklet", "Pied Kingfisher", "Prairie Warbler",
    "Black throated Sparrow", "Cape Glossy Starling", "Yellow Warbler",
    "Le Conte Sparrow", "Clark Nutcracker", "Caspian Tern",
    "Painted Bunting", "White Pelican", "Green Violetear",
    "Nashville Warbler", "Blue Jay", "White breasted Nuthatch",
    "Eared Grebe", "White breasted Kingfisher", "Pine Grosbeak",
    "Palm Warbler", "Scissor tailed Flycatcher", "Hooded Merganser",
    "Hooded Oriole", "Golden winged Warbler", "Bohemian Waxwing",
    "Cape May Warbler", "Spotted Catbird", "Western Meadowlark",
    "Yellow throated Vireo", "Nighthawk", "Heermann Gull",
    "Pied billed Grebe", "Mallard", "Red eyed Vireo",
    "Forsters Tern", "Fox Sparrow", "Brown Creeper",
    "Harris Sparrow", "Brown Thrasher", "Rhinoceros Auklet",
    "Savannah Sparrow", "Gray crowned Rosy Finch", "Carolina Wren",
    "Hooded Warbler", "Sayornis", "Bronzed Cowbird",
    "Cedar Waxwing", "Tree Swallow", "Bobolink",
    "Rufous Hummingbird", "Canada Warbler", "Summer Tanager",
    "Whip poor Will", "Pine Warbler", "Barn Swallow",
    "Vesper Sparrow", "Eastern Towhee", "White throated Sparrow",
    "Horned Puffin", "Gray Kingbird", "Western Wood Pewee",
    "Loggerhead Shrike", "Blue Grosbeak", "Ivory Gull",
    "Green tailed Towhee", "Brown Pelican", "Warbling Vireo",
    "Green Jay", "Ovenbird", "Pacific Loon",
    "Red cockaded Woodpecker", "Cerulean Warbler", "Geococcyx",
    "Bay breasted Warbler", "Ruby throated Hummingbird",
    "Chuck will Widow", "Magnolia Warbler", "Nelson Sharp tailed Sparrow"
]

# Original dataset root and splits
orig_root = r"C:\arpit\project3\resnet_dataset"
splits = ["train", "test"]

# New dataset root
new_root = r"C:\arpit\project3\resnet_dataset_top100"

for split in splits:
    src_split_dir = os.path.join(orig_root, split)
    dst_split_dir = os.path.join(new_root, split)
    os.makedirs(dst_split_dir, exist_ok=True)

    # List all class folders in the source split
    src_folders = os.listdir(src_split_dir)

    for cls in selected_classes:
        # Build the folder-name suffix: replace spaces with underscores
        suffix = cls.replace(" ", "_")

        # Find the matching folder in the source
        match = next(
            (f for f in src_folders if f.split('.', 1)[-1] == suffix),
            None
        )

        if match is None:
            print(f"⚠️  Could not find folder for class '{cls}' in {src_split_dir}")
            continue

        src_path = os.path.join(src_split_dir, match)
        dst_path = os.path.join(dst_split_dir, cls)  

        # Copy entire directory tree
        shutil.copytree(src_path, dst_path)
        print(f"Copied {src_path} → {dst_path}")

print("Done! New dataset with top‑100 classes created at:", new_root)
```
